```python
torch_rand1 = torch.rand(100, 100, 100, 100).to(device)
torch_rand2 = torch.rand(100, 100, 100, 100).to(device)
np_rand1 = torch.rand(100, 100, 100, 100)
np_rand2 = torch.rand(100, 100, 100, 100)

start_time = time.time()

rand = (torch_rand1 @ torch_rand2)

end_time = time.time()

elapsed_time = end_time - start_time
# gpu time to do a task
print(f"{elapsed_time:.8f}")


start_time = time.time()

rand = np.multiply(np_rand1, np_rand2)

end_time = time.time()

elapsed_time = end_time - start_time
# cpu time to do a task
print(f"{elapsed_time:.8f}")

# parallel proccessing is a lot quicker for a lot of easy tasks
```

```python
# torch.stack, torch.multinomial, torch.tril, torch.triu, input.T / input.transpose, nn.Linear, torch.cat, F.softmax (show all of the functions/methods with pytorch docs)

# define a probability tensor
probabilities = torch.tensor([0.1,0.9])
# 10% or 0.1 => 0, 90% or 0.9 => 1. each probability points to the idex of the probability in the tensor
# Draw 5 samples from the multinomial distribution
samples = torch.multinomial(probabilities, num_samples = 10, replacement = True)
print(samples)
# There is a 10% chance that we will get a 0 (becuase index 0) and a 90% chance that we will get a 1 (becuase index 1)
```

```python
# using neural netowrk to train parameters
import torch.nn as nn
sample = torch.tensor([10.,10.,10.])
# fully connected layer with: 3 input features, 3 output features, No bias (bias=False)
# This layer computes:output = 𝑊⋅input where W is a 3×3 weight matrix.
linear = nn.Linear(3, 3, bias=False)
print(linear(sample))
```

```python
# softmax function exponentiates the tensor and then divides by the the sum of the exponentiation
# exponentials are better because it Sharpens distinctions, is Always positive, Differentiable, Sensitive to differences, and Models uncertainty
import torch.nn.functional as F

# create a tensor
tensor1 = torch.tensor([1.0, 2.0, 3.0])

# apply softmax using torch.nn.functional.softmax()
softmax_output = F.softmax(tensor1, dim=0)

print(softmax_output)
```

```python
# Initialize an embedding layer
vocab_size = 1000
embedding_dim = 100
embedding = nn.Embedding(vocab_size, embedding_dim)

# Create some input indices
input_indices = torch.LongTensor([1, 5, 3, 2])

# Apply the embedding later
embedded_output = embedding(input_indices)

# The output will be a tensor of shape (4,100), where 4 is the number of inputs
# and 100 is the dimensionality of the embedding vectors
print(embedded_output.shape)
```

```python
# Multiplying two tensors together
a = torch.tensor([[1,2],[3,4],[5,6]])
b = torch.tensor([[7,8,9],[10,11,12]])
print(a @ b)
print(torch.matmul(a, b))
```

```python
# integers and floats cannot be multiplied in pytorch- need to cast ints
int_64 = torch.randint(1, (3, 2)).float()
# type int64
float_32 = torch.rand(2,3)
# type float32
result = torch.matmul(int_64, float_32)
print(result)
```

```python
# views allows us to get the vectors back out of a tensor
a = torch.rand(2, 3, 5)
print(a.shape)
x, y, z = a.shape
a = a.view(x, y, z)
print(x, y, z)
print(a.shape)
```

```python
x = torch.tensor([-0.05], dtype = torch.float32)
# applies the sigmoid activation function (output- between 0 and 1)
y = F.sigmoid(x)
print(y)
```

```python
x = torch.tensor([0], dtype = torch.float32)
# applies the tanh activation function (output- between -1 and 1)
y = F.tanh(x)
print(y)
```
